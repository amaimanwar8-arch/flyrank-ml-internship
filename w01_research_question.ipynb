## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

```python
print(""" Decision this improves: which pages a content reviewer looks at FIRST out of thousands,
given limited review capacity.

Unit of analysis: one page (one content_id).

Action someone takes: a reviewer opens the top-ranked pages and decides whether to
refresh, expand, protect, or leave a page alone.

Cost of a wrong call: if the ranking is wrong, the team spends limited review time on
pages that didn't need attention, while pages that were actually declining or losing
visibility get reviewed late or not at all. The cost isn't catastrophic — it's wasted
attention, not a broken product — but it's real, since review capacity is limited.

Why data/ML can help: a hand-written rule can only combine 2-3 signals before it gets
unreadable, but real pages have many interacting signals (position, freshness, CTR,
word count, trend). A model can weigh these together and rank pages by evidence instead
of a reviewer guessing which of thousands of pages to check first.""")
```

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
decline_rate = df["trend_direction"].str.lower().eq("down").mean()
print(f"1) {df.shape[0]:,} pages in the starter set; {decline_rate:.1%} are currently declining")

visible = df[df["impressions_90d"] >= 100]
stale_visible = ((df["days_since_last_update"] >= 180) & (df["impressions_90d"] >= 500)).sum()
print(f"2) {stale_visible:,} pages are both stale (no update in 180+ days) AND still getting real traffic (500+ impressions) - prime review candidates a reviewer can't easily spot by eye")

low_ctr_page1 = visible[(visible["position_tier"] == "page_1") & (visible["ctr"] < 0.20)].shape[0]
print(f"3) {low_ctr_page1:,} page-1-ranked pages have CTR under 0.20 despite ranking well - a signal not visible from position alone")

print("\nReference from the repo's own pipeline (outputs/model_results.json): a hand-written")
print("baseline rule hits Precision@50 = 0.240, while a random forest reaches 0.740 on the")
print("same task - about 3x more of the top 50 flagged pages are correct.")
```
